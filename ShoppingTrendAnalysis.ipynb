# Shopping Trends Analysis

Exploratory data analysis of customer shopping behavior using Python, Pandas, Matplotlib, and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

shop = pd.read_csv("shopping_trends_updated1.csv")
shop.head()
```

```python
# Basic data-quality checks
print("Shape:", shop.shape)
print("Missing values:")
print(shop.isnull().sum())
```

1 What is the overall distribution of customer ages in the dataset?

```python
# Age distribution
shop["Age"].describe()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(shop["Age"], bins=15, kde=True)
plt.title("Distribution of Customer Ages")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

2 How does the average purchase amount vary across different product categories?

```python
category_purchase = (
    shop.groupby("Category")["Purchase Amount (USD)"]
    .mean()
    .sort_values(ascending=False)
)
category_purchase
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(
    x=category_purchase.index,
    y=category_purchase.values
)
plt.title("Average Purchase Amount by Product Category")
plt.xlabel("Category")
plt.ylabel("Average Purchase Amount (USD)")
plt.xticks(rotation=20)
plt.show()
```

3 Which gender has the highest number of purchases?

```python
gender_counts = shop["Gender"].value_counts()
gender_counts
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(x=gender_counts.index, y=gender_counts.values)
plt.title("Number of Purchases by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Purchases")
plt.show()
```

4 What are the most commonly purchased items in each category?

```python
most_common_items = (
    shop.groupby(["Category", "Item Purchased"])
    .size()
    .reset_index(name="Purchase Count")
    .sort_values(["Category", "Purchase Count"], ascending=[True, False])
)

most_common_items.groupby("Category").head(1)
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(
    data=shop,
    y="Item Purchased",
    hue="Category",
    order=shop["Item Purchased"].value_counts().index
)
plt.title("Purchased Items by Category")
plt.xlabel("Number of Purchases")
plt.ylabel("Item Purchased")
plt.legend(title="Category")
plt.show()
```

5 Are there any specific seasons or months where customer spending is significantly higher?

```python
# The dataset contains Season but does not contain a month column.
season_spending = (
    shop.groupby("Season")["Purchase Amount (USD)"]
    .agg(["count", "mean", "sum"])
    .sort_values("mean", ascending=False)
)
season_spending
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(
    data=season_spending.reset_index(),
    x="Season",
    y="mean",
    ax=axes[0]
)
axes[0].set_title("Average Purchase Amount by Season")
axes[0].set_ylabel("Average Purchase Amount (USD)")
axes[0].tick_params(axis="x", rotation=20)

sns.barplot(
    data=season_spending.reset_index(),
    x="Season",
    y="sum",
    ax=axes[1]
)
axes[1].set_title("Total Spending by Season")
axes[1].set_ylabel("Total Purchase Amount (USD)")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```

10 How does the frequency of purchases vary across different age groups?

```python
shop["Age Group"] = pd.cut(
    shop["Age"],
    bins=[0, 18, 30, 50, 70, 100],
    labels=["Teen", "Young Adult", "Adult", "Middle-Aged", "Senior"]
)

age_frequency = pd.crosstab(
    shop["Age Group"],
    shop["Frequency of Purchases"]
)
age_frequency
```

```python
plt.figure(figsize=(11, 6))
sns.heatmap(age_frequency, annot=True, fmt="d", cmap="Blues")
plt.title("Purchase Frequency by Age Group")
plt.xlabel("Frequency of Purchases")
plt.ylabel("Age Group")
plt.show()
```

11 Are there any correlations between the size of the product and the purchase amount?

```python
# Size is categorical, so compare purchase amounts across sizes.
size_purchase = (
    shop.groupby("Size")["Purchase Amount (USD)"]
    .agg(["count", "mean", "median"])
    .sort_values("mean", ascending=False)
)
size_purchase
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=shop, x="Size", y="Purchase Amount (USD)")
plt.title("Purchase Amount Distribution by Product Size")
plt.xlabel("Product Size")
plt.ylabel("Purchase Amount (USD)")
plt.show()
```

12 Which shipping type is preferred by customers for different product categories?

```python
shipping_preference = (
    shop.groupby(["Category", "Shipping Type"])
    .size()
    .reset_index(name="Purchase Count")
    .sort_values(["Category", "Purchase Count"], ascending=[True, False])
)

shipping_preference.groupby("Category").head(1)
```

```python
plt.figure(figsize=(11, 6))
sns.countplot(
    data=shop,
    x="Category",
    hue="Shipping Type"
)
plt.title("Shipping Type Preference by Product Category")
plt.xlabel("Category")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=20)
plt.legend(title="Shipping Type", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
